# Quản Lý Kho Hàng Tự Động
File notebook này tự động tính toán số lượng **Tồn kho** dựa trên công thức: 
$$\text{Tồn} = \text{Nhập} - \text{Xuất}$$

```python
import pandas as pd

# 1. Khởi tạo dữ liệu nhập và xuất ban đầu
data = {
    'Tên Sản Phẩm': ['Sản phẩm A', 'Sản phẩm B', 'Sản phẩm C', 'Sản phẩm D'],
    'Nhập': [100, 250, 150, 80],
    'Xuất': [30, 120, 150, 0]
}

# 2. Tạo DataFrame
df = pd.DataFrame(data)

# 3. Tính toán cột 'Tồn' tự động theo từng dòng
df['Tồn'] = df['Nhập'] - df['Xuất']

# 4. Hiển thị bảng dữ liệu
df
```
